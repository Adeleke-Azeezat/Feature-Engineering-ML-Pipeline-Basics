Feature Engineering

Two new features are created for this analysis. Firstly, an average substance-use rate is calculated from alcohol use disorder, cocaine use,and tobacco use among individuals aged 18–25. This feature provides a combined measure of other substance-use patterns that may be associated with marijuana use.

Secondly, a young-adult population ratio is created by dividing the population aged 18–25 by the total population represented by the three age groups (12–17, 18–25, and 26+). This transforms the population counts into a proportion and provides information about the relative size of the 18–25 age group.



```python
df["Average_Other_Substance_Use_Rate_18_25"] = df[
    [
        "Rates.Alcohol.Use Disorder Past Year.18-25",
        "Rates.Illicit Drugs.Cocaine Used Past Year.18-25",
        "Rates.Tobacco.Use Past Month.18-25"
    ]
].mean(axis=1)
```

```python
df["Young_Adult_Population_Ratio"] = (
    df["Population.18-25"] /
    (
        df["Population.12-17"] +
        df["Population.18-25"] +
        df["Population.26+"]
    )
)
```

Define Features and Target

The target variable for this machine learning task is the rate of marijuana use in the past year among individuals aged 18–25.The selected features include state, year, young adult popuulation ratio, alcohol use disorder, cocaine use, and tobacco use among the same age group.


```python
features = [
    "State",
    "Year",
    "Young_Adult_Population_Ratio",
    "Rates.Alcohol.Use Disorder Past Year.18-25",
    "Rates.Illicit Drugs.Cocaine Used Past Year.18-25",
    "Rates.Tobacco.Use Past Month.18-25"
]

X = df[features]
y = df["Rates.Marijuana.Used Past Year.18-25"]
```

```python
print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

Train/Test Split
The dataset is divided into training and testing sets using an 80/20 split respectively. The training data will be used to learn the preprocessing parameters and train the machine learning model, while the test data will be reserved for final evaluation.
A random state of 42 is used to ensure that the same split can be reproduced.

```python
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.20,random_state=42)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

Identify Feature Types
The selected features are divided into numerical and categorical variables.
Numerical features include year, young adult population, alcohol use disorder, cocaine use, and tobacco use. These variables require numerical preprocessing.
State is treated as a categorical feature because it represents differentstate categories rather than a continuous numerical measurement. It will be encoded using one-hot encoding. 

```python
numeric_features = [
    "Year",
    "Young_Adult_Population_Ratio",
    "Rates.Alcohol.Use Disorder Past Year.18-25",
    "Rates.Illicit Drugs.Cocaine Used Past Year.18-25",
    "Rates.Tobacco.Use Past Month.18-25"
]

categorical_features = ["State"]
```

Numerical Feature Preprocessing

The numerical features are processed using a two-step pipeline. The missing numerical values are replaced with the median value (it is less sensitive to outliers) of the corresponding feature,then standardScaler is used to standardize the numerical variables so that they are placed on a comparable scale.

The preprocessing steps are fitted using the training data to avoid data leakage.

```python
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

Categorical Feature Preprocessing

The State variable is categorical, so OneHotEncoder is used to transform each state into numerical binary features.The missing values are replaced with the most frequent state and the 'handle_unknown="ignore"' option ensures that the preprocessing pipeline does not produce an error if a category appears in the test data that was not present in the training data.

```python
categorical_transformer = Pipeline(steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

Combine Preprocessing Steps

ColumnTransformer is used to apply the appropriate preprocessing method to each type of feature.
Numerical variables are passed through the numerical preprocessing pipeline, and the State variable is passed through the categorical pipeline, where it is one-hot encoded.

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

print("Original Training Data Shape:", X_train.shape)
print("Processed Training Data Shape:", X_train_processed.shape)
```

```python
X_test_processed = preprocessor.transform(X_test)

print("Original Testing Data Shape:", X_test.shape)
print("Processed Testing Data Shape:", X_test_processed.shape)
```

The machine learning pipeline is trained using the training dataset. The pipeline contains both the preprocessing steps and the Linear Regression model. Using the pipeline ensures that the preprocessing operations are fitted only on the training data. This helps prevent data leakage because information from the test dataset is not used during the training process.


```python
from sklearn.linear_model import LinearRegression

model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])
```

```python
model.fit(X_train, y_train)
```

A 5-fold cross-validation is performed on the training dataset, the training data is divided into five subsets,and the model is trained on four folds and validated on the remaining fold. This process is repeated five times so that each fold is used as the validation set once.

This helps determine whether the model performs consistently across different subsets of the training data and reduces the possibility of relying on a single train-validation split. The test dataset remains untouched during cross-validation and is reserved for the final evaluation of the trained model.

```python
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    model,
    X_train,
    y_train,
    cv=5,
    scoring="r2"
)

print("Cross-validation scores:", cv_scores)
print("Mean CV score:", cv_scores.mean())
```

```python
y_pred = model.predict(X_test)
```

```python
results = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})
results.head()
```

Model Evaluation
The trained model is evaluated using the test dataset.The evaluation metrics used are Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R² Score.


```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R²:", r2)
```

```python
print("Mean CV R²:", cv_scores.mean())
print("Test R²:", r2)
```

Overall, the model demonstrates a relatively strong predictive performance based on the R² score. However, the MAE and RMSE indicate that some prediction errors remain.The model performed similarly during cross-validation and on the test set. The small difference suggests that the model's performance is fairly consistent and that there isn't a major indication of overfitting.
